```python
knowledge_df = pd.DataFrame(financial_knowledge)

knowledge_df
```

```python
def search_knowledge(keyword):

    results = knowledge_df[
        knowledge_df["keyword"].str.contains(keyword, case=False, na=False)
    ]

    if results.empty:
        print("No relevant financial information found.")
    else:
        for index, row in results.iterrows():
            print("\nTopic:", row["topic"])
            print("Information:", row["information"])
```

```python
search_knowledge("credit score")
```

```python
search_knowledge("income")
search_knowledge("debt")
search_knowledge("repayment")
```

```python
print("Available Financial Topics:\n")

for topic in knowledge_df["topic"]:
    print("-", topic)
```
