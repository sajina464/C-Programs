```python
import pandas as pd
```

```python
credit_results = pd.DataFrame({
    "Applicant": ["A001", "A002", "A003", "A004"],
    "Credit_Score": [580, 720, 620, 780],
    "Income": [25000, 60000, 30000, 80000],
    "Loan_Amount": [300000, 250000, 400000, 300000],
    "Risk": ["High Risk", "Low Risk", "High Risk", "Low Risk"]
})

print(credit_results)
```

```python
financial_knowledge = {
    "High Risk": [
        "A low credit score may indicate a weaker credit history.",
        "A high loan amount compared with income can increase repayment risk.",
        "Paying bills and existing loans on time can help maintain a healthy credit profile.",
        "Reducing existing debt can improve financial stability."
    ],

    "Low Risk": [
        "A good credit score generally indicates a stronger credit history.",
        "Stable income can support the ability to repay a loan.",
        "Maintaining timely payments can help preserve a healthy credit profile."
    ]
}
```

```python
def connect_financial_knowledge(risk):

    print("\nCredit Risk:", risk)
    print("Financial Knowledge:")

    for information in financial_knowledge[risk]:
        print("-", information)
```

```python
for index, row in credit_results.iterrows():

    print("\n==============================")
    print("Applicant:", row["Applicant"])
    print("Credit Score:", row["Credit_Score"])
    print("Income:", row["Income"])
    print("Loan Amount:", row["Loan_Amount"])

    connect_financial_knowledge(row["Risk"])
```

```python
def financial_advisor(row):

    print("\n--------------------------------")
    print("Applicant:", row["Applicant"])
    print("Risk Level:", row["Risk"])

    reasons = []

    if row["Credit_Score"] < 650:
        reasons.append(
            "Credit score is relatively low. Maintaining timely payments can help improve credit history."
        )

    if row["Income"] < 35000:
        reasons.append(
            "Income is relatively low. Stable income can support loan repayment."
        )

    if row["Loan_Amount"] > row["Income"] * 8:
        reasons.append(
            "Loan amount is high compared with income. A lower debt burden can reduce repayment risk."
        )

    if not reasons:
        reasons.append(
            "The available financial indicators show a relatively lower-risk profile."
        )

    print("\nFinancial Explanation:")

    for reason in reasons:
        print("-", reason)
```

```python
for index, row in credit_results.iterrows():
    financial_advisor(row)
```
