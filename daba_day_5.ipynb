```python
import pandas as pd
```

```python
applicant = {
    "name": "Applicant A001",
    "credit_score": 580,
    "income": 25000,
    "loan_amount": 300000,
    "employment_years": 1,
    "risk": "High Risk"
}

print("Applicant:", applicant["name"])
print("Credit Score:", applicant["credit_score"])
print("Income:", applicant["income"])
print("Loan Amount:", applicant["loan_amount"])
print("Employment:", applicant["employment_years"], "years")
print("Risk Level:", applicant["risk"])
```

```python
def create_risk_prompt(applicant):

    prompt = f"""
You are a financial education assistant.

Explain the credit-risk result in simple language.

Applicant Details:
- Credit Score: {applicant['credit_score']}
- Monthly Income: ₹{applicant['income']}
- Loan Amount: ₹{applicant['loan_amount']}
- Employment Experience: {applicant['employment_years']} years
- Credit Risk: {applicant['risk']}

Instructions:
1. Explain the risk level clearly.
2. Identify the main factors that may have contributed to the result.
3. Do not make unsupported assumptions.
4. Use simple and understandable language.
5. Do not guarantee loan approval or rejection.
6. Clearly state that the result is an educational model prediction.

Give the response in this format:

Risk Explanation:
Possible Factors:
Summary:
"""

    return prompt


risk_prompt = create_risk_prompt(applicant)

print(risk_prompt)
```

```python
def create_guidance_prompt(applicant):

    prompt = f"""
You are a financial education assistant.

Provide general financial guidance based on the applicant's
credit-risk information.

Applicant Details:
- Credit Score: {applicant['credit_score']}
- Monthly Income: ₹{applicant['income']}
- Loan Amount: ₹{applicant['loan_amount']}
- Employment Experience: {applicant['employment_years']} years
- Risk Level: {applicant['risk']}

Provide practical and responsible guidance about:

1. Managing existing debt
2. Making payments on time
3. Maintaining a healthy credit history
4. Managing income and expenses
5. Considering an affordable loan amount

Rules:
- Use simple language.
- Do not promise a particular financial outcome.
- Do not invent financial information.
- Do not make the decision for the applicant.
- Mention that this is general educational guidance.

Output format:

Financial Guidance:
1.
2.
3.
4.
5.

Important Note:
"""

    return prompt


guidance_prompt = create_guidance_prompt(applicant)

print(guidance_prompt)
```

```python
financial_knowledge = [
    "A lower credit score may indicate a weaker credit history.",
    "Stable income can help demonstrate the ability to repay a loan.",
    "A large loan amount compared with income may increase repayment risk.",
    "Making payments on time can help maintain a healthy credit history.",
    "Reducing existing debt can improve financial stability."
]

def create_rag_prompt(applicant, knowledge):

    knowledge_text = "\n".join(
        "- " + item for item in knowledge
    )

    prompt = f"""
You are a financial education assistant.

Use the applicant information and the retrieved financial
knowledge to explain the credit-risk result.

APPLICANT INFORMATION
---------------------
Credit Score: {applicant['credit_score']}
Monthly Income: ₹{applicant['income']}
Loan Amount: ₹{applicant['loan_amount']}
Employment Experience: {applicant['employment_years']} years
Risk Level: {applicant['risk']}

RETRIEVED FINANCIAL KNOWLEDGE
-----------------------------
{knowledge_text}

TASK
----
Explain the credit-risk result using only the information provided.

Include:

1. Credit-risk explanation
2. Possible contributing factors
3. Relevant financial knowledge
4. General financial guidance
5. Important limitations

Rules:
- Use simple language.
- Do not invent facts.
- Do not guarantee loan approval or rejection.
- Do not provide personalized investment advice.
- Clearly distinguish model prediction from actual lending decisions.
- Mention that the information is for educational purposes.

FINAL RESPONSE
--------------
Risk Explanation:
Possible Factors:
Relevant Financial Knowledge:
General Financial Guidance:
Limitations:
"""

    return prompt


rag_prompt = create_rag_prompt(
    applicant,
    financial_knowledge
)

print(rag_prompt)
```
