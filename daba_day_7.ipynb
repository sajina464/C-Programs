```python

# Financial RAG Assistant
# Testing Financial Queries with Different Credit Profiles

import pandas as pd

# 1. Sample credit profiles
credit_profiles = {
    "Excellent": {
        "credit_score": 820,
        "monthly_income": 80000,
        "monthly_debt": 12000,
        "credit_utilization": 15
    },
    "Good": {
        "credit_score": 750,
        "monthly_income": 60000,
        "monthly_debt": 18000,
        "credit_utilization": 25
    },
    "Fair": {
        "credit_score": 680,
        "monthly_income": 40000,
        "monthly_debt": 16000,
        "credit_utilization": 45
    },
    "Poor": {
        "credit_score": 580,
        "monthly_income": 30000,
        "monthly_debt": 18000,
        "credit_utilization": 75
    }
}

# 2. Financial queries to test
financial_queries = [
    "What is my credit profile?",
    "Am I ready to apply for a loan?",
    "How can I improve my credit score?",
    "Is my credit utilization healthy?",
    "How much of my income goes towards debt?"
]


# 3. Function to answer a financial query
def financial_rag_answer(query, profile_name, profile):
    score = profile["credit_score"]
    income = profile["monthly_income"]
    debt = profile["monthly_debt"]
    utilization = profile["credit_utilization"]

    debt_ratio = (debt / income) * 100

    # Basic profile assessment
    if score >= 750:
        rating = "Excellent"
    elif score >= 700:
        rating = "Good"
    elif score >= 650:
        rating = "Fair"
    else:
        rating = "Needs improvement"

    query_lower = query.lower()

    if "credit profile" in query_lower:
        answer = (
            f"Your credit score is {score}. "
            f"Profile assessment: {rating}. "
            f"Monthly income: Rs. {income:,}. "
            f"Monthly debt: Rs. {debt:,}."
        )

    elif "loan" in query_lower:
        if score >= 750 and debt_ratio < 40:
            answer = (
                "Your profile has some positive indicators for a loan "
                "application. Approval is not guaranteed."
            )
        elif score >= 650 and debt_ratio < 50:
            answer = (
                "You may want to review your debt and repayment history "
                "before applying. Lender criteria will vary."
            )
        else:
            answer = (
                "Consider reducing existing debt and improving your "
                "repayment history before applying. Approval depends "
                "on the lender and other eligibility requirements."
            )

    elif "improve" in query_lower:
        answer = (
            "Pay EMIs and credit card bills on time, reduce outstanding "
            "balances, avoid unnecessary loan applications, and check "
            "your credit report for errors."
        )

    elif "utilization" in query_lower:
        if utilization <= 30:
            answer = (
                f"Your utilization is {utilization}%, which is generally "
                "a reasonable target. Lower utilization may help."
            )
        else:
            answer = (
                f"Your utilization is {utilization}%. Consider reducing "
                "your outstanding credit card balance. Around 30% or "
                "less is a commonly used guideline, not a guarantee."
            )

    elif "income" in query_lower or "debt" in query_lower:
        answer = (
            f"{debt_ratio:.1f}% of your monthly income goes towards "
            f"the listed monthly debt payments "
            f"(Rs. {debt:,} / Rs. {income:,})."
        )

    else:
        answer = (
            "Please ask about your credit profile, loan readiness, "
            "credit score improvement, utilization, or debt ratio."
        )

    return {
        "profile": profile_name,
        "query": query,
        "credit_score": score,
        "debt_ratio_percent": round(debt_ratio, 2),
        "answer": answer
    }


# 4. Test every query against every credit profile
results = []

for profile_name, profile in credit_profiles.items():
    for query in financial_queries:
        result = financial_rag_answer(
            query, profile_name, profile
        )
        results.append(result)

# 5. Display test results
results_df = pd.DataFrame(results)

print("Total test cases:", len(results_df))
display(results_df)

# 6. Save test results for GitHub / project report
results_df.to_csv(
    "financial_query_test_results.csv",
    index=False
)

print("Testing completed successfully!")
print("CSV report saved as financial_query_test_results.csv")
```
